# Descriptive statistics of the campaign dataset

**Revision note (v2.0):** This notebook computes the statistics of the authoritative dataset `data/processed/full_data.csv` (N = 145,784) used by every analysis notebook. All statistical claims in the main text and Supplemental Material are updated from this output.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../../data/processed/full_data.csv')
print(f"N = {len(df)}")

cols = {
    'temperature (degC)': df['temperature'],
    'humidity (%RH)':     df['humidity'],
    'pressure (hPa)':     df['pressure'],
}

print()
print("=== Descriptive statistics ===")
stats = {}
for name, s in cols.items():
    stats[name] = {
        'mean': s.mean(),
        'std':  s.std(ddof=0),
        'min':  s.min(),
        'max':  s.max(),
    }
    print(f"{name:<20} mean={s.mean():10.3f}  std={s.std(ddof=0):6.3f}  "
          f"min={s.min():8.2f}  max={s.max():8.2f}")

# --- statistics used in the manuscript text ---
T_C = df['temperature'].values
P_hPa = df['pressure'].values
H_pct = df['humidity'].values

print()
print("=== Values quoted in the manuscript / SM ===")
print(f"mean T   = {T_C.mean():.1f} °C   = {T_C.mean()+273.15:.1f} K")
print(f"mean RH  = {H_pct.mean():.1f} %")
print(f"mean P   = {P_hPa.mean():.1f} hPa")
print(f"std T    = {T_C.std(ddof=0):.1f} °C")
print(f"std RH   = {H_pct.std(ddof=0):.1f} %")
print(f"std P    = {P_hPa.std(ddof=0):.1f} hPa")
print(f"T range  = [{T_C.min():.1f}, {T_C.max():.1f}] °C")
print(f"RH range = [{H_pct.min():.1f}, {H_pct.max():.1f}] %")
print(f"P range  = [{P_hPa.min():.1f}, {P_hPa.max():.1f}] hPa")
print()
print("=== Mean conditions for Kramers-Kronig (SM S5) ===")
print(f"T_mean = {T_C.mean()+273.15:.1f} K   (use 302 K)")
print(f"P_mean = {P_hPa.mean():.1f} hPa     (use 986 hPa)")
print(f"RH_mean = {H_pct.mean():.1f} %")

# also count fraction above 25 C for cross-check
frac_above25 = np.mean(T_C > 25) * 100
print(f"Fraction of observations above 25 °C: {frac_above25:.1f} % (cross-check vs SM S2)")
```

```python
import json
mean_conditions = {
    'N': len(df),
    'T_mean_C': float(T_C.mean()),
    'T_mean_K': float(T_C.mean() + 273.15),
    'RH_mean': float(H_pct.mean()),
    'P_mean_hPa': float(P_hPa.mean()),
    'T_std_C': float(T_C.std(ddof=0)),
    'RH_std': float(H_pct.std(ddof=0)),
    'P_std_hPa': float(P_hPa.std(ddof=0)),
    'T_min_C': float(T_C.min()), 'T_max_C': float(T_C.max()),
    'RH_min': float(H_pct.min()), 'RH_max': float(H_pct.max()),
    'P_min_hPa': float(P_hPa.min()), 'P_max_hPa': float(P_hPa.max()),
}
with open('campaign_statistics.json', 'w') as f:
    json.dump(mean_conditions, f, indent=2)
print("Saved campaign_statistics.json")
```
